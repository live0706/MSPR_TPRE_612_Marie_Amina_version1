# EDA et modelisation CO2 - TPRE622

Notebook d'analyse exploratoire pour le sujet ObRail IA CO2.

```python
from pathlib import Path
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

DATASET = Path('../data/processed/trips_cleaned_final.csv')
df = pd.read_csv(DATASET)
df.head()
```

## Qualite des donnees

```python
quality = pd.DataFrame({
    'column': df.columns,
    'dtype': [str(df[col].dtype) for col in df.columns],
    'missing_rate': [df[col].isna().mean() for col in df.columns],
    'unique_values': [df[col].nunique(dropna=True) for col in df.columns],
})
quality
```

## Variables retenues

```python
variables = pd.DataFrame([
    {'variable': 'distance_km', 'role': 'feature', 'type': 'numerique'},
    {'variable': 'operator_name', 'role': 'feature', 'type': 'categorielle'},
    {'variable': 'service_type', 'role': 'feature', 'type': 'categorielle'},
    {'variable': 'train_type', 'role': 'feature', 'type': 'categorielle'},
    {'variable': 'country', 'role': 'feature', 'type': 'categorielle'},
    {'variable': 'source_origin', 'role': 'feature', 'type': 'categorielle'},
    {'variable': 'departure_time', 'role': 'feature engineering', 'type': 'temporelle'},
    {'variable': 'arrival_time', 'role': 'feature engineering', 'type': 'temporelle'},
    {'variable': 'co2_emissions', 'role': 'target', 'type': 'numerique'},
])
variables
```

## Distributions et correlations

```python
sample = df[['distance_km', 'co2_emissions', 'service_type', 'country']].dropna().sample(min(10000, len(df)), random_state=42)
sns.scatterplot(data=sample, x='distance_km', y='co2_emissions', hue='service_type', alpha=0.35)
plt.title('Distance vs emissions CO2')
plt.show()
```
